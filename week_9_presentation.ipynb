```python
import pandas as pd

pokemon = pd.read_csv('Pokemon.csv')
pokemon
```

## Question \#1
What is the relationship between Pokemon type and Attack?

```python
df1 = pokemon[['Type 1', 'Attack']].copy()
df1.rename(columns={'Type 1': 'Type_1'}, inplace=True)
df1 = df1.groupby('Type_1')['Attack'].mean().sort_values(ascending=False)
df1.plot(kind='bar')
```

## Question \#2
Does the mean Attack power increase from Generation 1 through Generation 6?

```python
df2 = pokemon[['Generation', 'Attack']].copy()
df2.groupby('Generation')['Attack'].mean().plot(kind='line')
df2.plot(kind='bar')
```
